#### Data Cleaning

```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/car_price_dataset.csv")

df.head()
```

```python
#Remove the Unnamed column
df.drop("Unnamed: 0", axis=1, inplace=True)
```

```python
df.columns.to_list()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
print("Shape after removing duplicates:", df.shape)
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
print("Mileage:")
print(df['mileage'].head(10))

print("\nEngine:")
print(df['engine'].head(10))
```
